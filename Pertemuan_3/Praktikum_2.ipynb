## Langkah 0 - Load Library

```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, StandardScaler
```

## Langkah 1 - Load Data

```python
dpath = 'Titanic-Dataset.csv'
df = pd.read_csv(dpath)
df.head()
```

## Langkah 2 - Pre Processing Encoding Sex

```python
# --- 2. Encoding fitur 'Sex' ---
df['Sex'] = df['Sex'].map({'male': 1, 'female': 0})
print("=== Hasil Setelah Encoding 'Sex' ===")
print(df[['Sex']].head(), "\n")
```

## Langkah 3 - Verifikasi hasil encoding "Sex"

```python
print(df[['Sex']].head(), "\n")
```

## Langkah 4 - Pre Processing Encoding "Cabin"

## Langkah 5 - Verifikasi Hasil encoding "Cabin"

```python
# Mengisi semua nilai missing sekaligus pada dataframe utama
df.fillna({
    'Age': df['Age'].mean(),
    'Cabin': 'Unknown',
    'Embarked': df['Embarked'].mode()[0]
}, inplace=True)

# --- Encoding fitur 'Cabin' ---
print("=== Hasil Setelah Fill NaN pada 'Cabin' ===")
print(df[['Cabin']].head(5), "\n")

# Mengambil huruf pertama Cabin sebagai Deck
df['Cabin'] = df['Cabin'].apply(
    lambda x: x[0] if x != 'Unknown' else 'U'
)

print("=== Hasil Ambil huruf pertama 'Cabin' (Deck) ===")
print(df[['Cabin']].head(5), "\n")
```

```python
df.isnull().sum()
```

## Langkah 6 - Preprocessing Imputasi Data

```python
# --- 6. Standarisasi fitur 'Age' ---
df['Age'] = df[['Age']].fillna(df['Age'].median())
print("=== Hasil Setelah Imputasi Data pada Usia dengan median ===")
print(df[['Age']].head(10), "\n")
```

## Langkah 7 - Preprocessing Standardisasi

## Langlah 8 - Verifikasi Hasil Standarisasi

```python
scaler = StandardScaler()
df['Age'] = scaler.fit_transform(df[['Age']])
print("=== Hasil Setelah Standardisasi 'Age' ===")
print(df[['Age']].head(10), "\n")
```
