## 1. Pisahkan antara variabel yang dapat digunakan dan variabel yang tidak dapat digunakan.


```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, StandardScaler
```

```python
# Read data
df = pd.read_csv('burnout_dataset1.csv')

df.head()
```

```python
# 1. Pemisahan Variabel
df_clean = df.drop(columns=['Person_ID', 'AI_Wellness_Recommendation'], errors='ignore')
df_clean = df_clean.fillna(df_clean.mean(numeric_only=True))

df_clean.info()
```

## 2. akukan proses encoding pada kolom "Burnout_Risk".


```python
# 2. Encoding pada Kolom 'Burnout_Risk'
le = LabelEncoder()
df_clean['Burnout_Risk'] = le.fit_transform(df_clean['Burnout_Risk'])

cat_cols = df_clean.select_dtypes(include=['object']).columns
for col in cat_cols:
    df_clean[col] = le.fit_transform(df_clean[col].astype(str))

df_clean.head()
```

## 3. Lakukan proses standardisasi pada semua kolom yang memiliki nilai numerik


```python
# 3. Standardisasi Kolom Numerik
# Memisahkan nama kolom numerik selain target 'Burnout_Risk'
numeric_features = df_clean.columns.drop('Burnout_Risk')

# Mengimplementasikan StandardScaler
scaler = StandardScaler()
df_clean[numeric_features] = scaler.fit_transform(df_clean[numeric_features])

df_clean.head()
```
