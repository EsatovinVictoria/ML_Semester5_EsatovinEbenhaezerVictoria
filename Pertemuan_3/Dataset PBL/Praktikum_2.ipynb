## Langkah 0 - Load Library


```python
import pandas as pd
from sklearn.preprocessing import LabelEncoder, StandardScaler
```

## Langkah 1 - Load Data


```python
dpath = 'burnout_dataset1.csv'
df = pd.read_csv(dpath)
df.head()
```

## Langkah 2 - Pre Processing Encoding Gender


```python
# --- 2. Encoding fitur 'Gender' ---
df['Gender'] = df['Gender'].map({'Male': 1, 'Female': 0})
print("=== Hasil Setelah Encoding 'Gender' ===")
print(df[['Gender']].head(), "\n")
```

## Langkah 3 - Verifikasi hasil encoding "Gender"


```python
print(df[['Gender']].head(), "\n")
```

## Langkah 4 - Pre Processing Encoding "Stress_Level"


## Langkah 5 - Verifikasi Hasil encoding "Stress_Level"


```python
# Mengisi semua nilai missing sekaligus pada dataframe utama
df.fillna(df.mean(numeric_only=True), inplace=True)

# --- Encoding fitur 'Stress_Level' ---
print("=== Hasil Sebelum Encoding pada 'Stress_Level' ===")
print(df[['Stress_Level']].head(5), "\n")

# Mapping nilai Stress_Level ke numerik ordinal
stress_map = {'Low': 0, 'Moderate': 1, 'High': 2}
df['Stress_Level'] = df['Stress_Level'].map(stress_map)

print("=== Hasil Setelah Encoding 'Stress_Level' ===")
print(df[['Stress_Level']].head(5), "\n")
```

```python
df.isnull().sum()
```

## Langkah 6 - Preprocessing Imputasi Data


```python
# --- 6. Standarisasi fitur 'Age' ---
df['Age'] = df[['Age']].fillna(df['Age'].median())
print("=== Hasil Setelah Imputasi Data pada Usia dengan median ===")
print(df[['Age']].head(10), "\n")
```

## Langkah 7 - Preprocessing Standardisasi

## Langlah 8 - Verifikasi Hasil Standarisasi


```python
scaler = StandardScaler()
df['Age'] = scaler.fit_transform(df[['Age']])
print("=== Hasil Setelah Standardisasi 'Age' ===")
print(df[['Age']].head(10), "\n")
```

## Langkah 8 - Hasil setelah Preprocessing


```python
df.head()
```

## Langkah 9 - Seleksi Fitur


```python
df['Burnout_Risk'] = df['Burnout_Risk'].map({'Low': 0, 'Moderate': 1, 'High': 2})
df = df[['Burnout_Risk', 'Stress_Level', 'Age', 'Gender', 'Work_Hours_Per_Week']]
df.to_csv('burnout_dataset1-selected.csv', index=False)
df.head()
```
