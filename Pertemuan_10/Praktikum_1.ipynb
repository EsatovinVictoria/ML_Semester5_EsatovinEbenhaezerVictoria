# Praktikum 1

```python
import numpy as np

# Dataset XOR
X = np.array([[0,0],[0,1],[1,0],[1,1]])
y = np.array([[0],[1],[1],[0]])

# Parameter
input_size = 2
hidden_size = 2
output_size = 1
lr = 0.1

# Inisialisasi bobot
np.random.seed(42)
W1 = np.random.randn(input_size, hidden_size)
b1 = np.zeros((1, hidden_size))
W2 = np.random.randn(hidden_size, output_size)
b2 = np.zeros((1, output_size))

# Fungsi aktivasi
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def sigmoid_derivative(x):
    return x * (1 - x)

# Training
for epoch in range(10000):
    # Forward pass
    z1 = np.dot(X, W1) + b1
    a1 = sigmoid(z1)
    z2 = np.dot(a1, W2) + b2
    a2 = sigmoid(z2)

    # Hitung error
    error = y - a2

    # Backpropagation
    d_a2 = error * sigmoid_derivative(a2)
    d_W2 = np.dot(a1.T, d_a2)
    d_b2 = np.sum(d_a2, axis=0, keepdims=True)

    d_a1 = np.dot(d_a2, W2.T) * sigmoid_derivative(a1)
    d_W1 = np.dot(X.T, d_a1)
    d_b1 = np.sum(d_a1, axis=0, keepdims=True)

    # Update bobot
    W1 += lr * d_W1
    b1 += lr * d_b1
    W2 += lr * d_W2
    b2 += lr * d_b2

    if epoch % 1000 == 0:
        loss = np.mean(np.square(error))
        print(f"Epoch {epoch}, Loss: {loss}")

# Output akhir
loss_awal = np.mean(np.square(error))
pred_awal = a2.copy()
print("Prediksi:")
print(a2)
```

# Tugas 1

```python
input_size = 2
hidden_size = 3
output_size = 1
lr = 0.1

np.random.seed(42)
W1_h3 = np.random.randn(input_size, hidden_size)
b1_h3 = np.zeros((1, hidden_size))
W2_h3 = np.random.randn(hidden_size, output_size)
b2_h3 = np.zeros((1, output_size))

for epoch in range(10000):
    z1 = np.dot(X, W1_h3) + b1_h3
    a1 = sigmoid(z1)
    z2 = np.dot(a1, W2_h3) + b2_h3
    a2 = sigmoid(z2)

    error = y - a2

    d_a2 = error * sigmoid_derivative(a2)
    d_W2 = np.dot(a1.T, d_a2)
    d_b2 = np.sum(d_a2, axis=0, keepdims=True)

    d_a1 = np.dot(d_a2, W2_h3.T) * sigmoid_derivative(a1)
    d_W1 = np.dot(X.T, d_a1)
    d_b1 = np.sum(d_a1, axis=0, keepdims=True)

    W1_h3 += lr * d_W1
    b1_h3 += lr * d_b1
    W2_h3 += lr * d_W2
    b2_h3 += lr * d_b2

    if epoch % 1000 == 0:
        loss = np.mean(np.square(error))
        print(f"Epoch {epoch}, Loss: {loss}")

loss_h3_sigmoid = np.mean(np.square(error))
pred_h3_sigmoid = a2.copy()
print("Prediksi (Hidden 3 - Sigmoid):")
print(a2)
```

```python
def relu(x):
    return np.maximum(0, x)

def relu_derivative(x):
    return (x > 0).astype(float)

input_size = 2
hidden_size = 3
output_size = 1
lr = 0.1

np.random.seed(42)
W1_relu = np.random.randn(input_size, hidden_size)
b1_relu = np.zeros((1, hidden_size))
W2_relu = np.random.randn(hidden_size, output_size)
b2_relu = np.zeros((1, output_size))

for epoch in range(10000):
    z1 = np.dot(X, W1_relu) + b1_relu
    a1 = relu(z1)
    z2 = np.dot(a1, W2_relu) + b2_relu
    a2 = sigmoid(z2)

    error = y - a2

    d_a2 = error * sigmoid_derivative(a2)
    d_W2 = np.dot(a1.T, d_a2)
    d_b2 = np.sum(d_a2, axis=0, keepdims=True)

    d_a1 = np.dot(d_a2, W2_relu.T) * relu_derivative(z1)
    d_W1 = np.dot(X.T, d_a1)
    d_b1 = np.sum(d_a1, axis=0, keepdims=True)

    W1_relu += lr * d_W1
    b1_relu += lr * d_b1
    W2_relu += lr * d_W2
    b2_relu += lr * d_b2

    if epoch % 1000 == 0:
        loss = np.mean(np.square(error))
        print(f"Epoch {epoch}, Loss: {loss}")

loss_h3_relu = np.mean(np.square(error))
pred_h3_relu = a2.copy()
print("Prediksi (Hidden 3 - ReLU):")
print(a2)
```

```python
import pandas as pd

df_hasil = pd.DataFrame({
    'Konfigurasi': ['Awal (Hidden 2 - Sigmoid)', 'Tugas 1 (Hidden 3 - Sigmoid)', 'Tugas 1 (Hidden 3 - ReLU)'],
    'Hidden Size': [2, 3, 3],
    'Aktivasi Hidden': ['Sigmoid', 'Sigmoid', 'ReLU'],
    'Final Loss': [loss_awal, loss_h3_sigmoid, loss_h3_relu]
})
df_hasil
```

Berdasarkan hasil pengujian di atas:

1. Ketika jumlah neuron pada hidden layer diubah dari 2 menjadi 3 dengan fungsi aktivasi Sigmoid, nilai loss akhir mengalami penurunan dari sekitar 0.02 menjadi 0.0023. Hal ini menunjukkan bahwa penambahan neuron memberikan ruang representasi yang lebih baik bagi model untuk memisahkan data XOR yang bersifat non-linear.

2. Penggunaan fungsi aktivasi ReLU pada hidden layer menghasilkan loss akhir yang lebih kecil lagi, yaitu sekitar 0.00035. Hal ini terjadi karena ReLU tidak mengalami masalah vanishing gradient seperti pada Sigmoid, sehingga proses pembaruan bobot saat backpropagation dapat berjalan lebih cepat dan efektif.
