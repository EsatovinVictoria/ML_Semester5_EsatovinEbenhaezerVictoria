## Langkah 1 - Load Data


```python
import pandas as pd

dpath = 'burnout_dataset1-fixed.csv'
df = pd.read_csv(dpath)
df.head()
```

## Langkah 2 - Slice Data


```python
df = df[['Burnout_Risk', 'Gender', 'Remote_Work', 'Stress_Level', 'Age', 'Work_Hours_Per_Week']]
df.head()
```

## Langkah 3 - Encoding


```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder() # membuat objek dari LabelEncoder
df['Gender'] = le.fit_transform(df['Gender']) # proses encoding
df['Remote_Work'] = le.fit_transform(df['Remote_Work']) # proses encoding
df['Stress_Level'] = le.fit_transform(df['Stress_Level']) # proses encoding
df['Burnout_Risk'] = le.fit_transform(df['Burnout_Risk']) # proses encoding
```

## Langkah 4 - Verifikasi Hasil


```python
df.head()
```

## Langkah 5 - Standardisasi


```python
from sklearn.preprocessing import StandardScaler

std = StandardScaler()
df['Age'] = std.fit_transform(df[['Age']])
df['Work_Hours_Per_Week'] = std.fit_transform(df[['Work_Hours_Per_Week']])
```
