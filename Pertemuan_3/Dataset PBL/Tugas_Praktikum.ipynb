## Anti-Burnout Dataset


### 1. Pisahkan antara variabel yang dapat digunakan dan variabel yang tidak dapat digunakan.


```python
# Load dataset
import pandas as pd

df = pd.read_csv('burnout_dataset1.csv')

df.head()

# Struktur dataset
print(f'Jumlah data: {df.shape[0]}')
print(f'Jumlah kolom: {df.shape[1]}')
print(df.info())
```

```python
# Memisahkan variabel yang dapat digunakan dengan yang tidak dapat digunakan
df = df.drop(columns=['Person_ID', 'AI_Wellness_Recommendation'], errors='ignore')
df = df.fillna(df.mean(numeric_only=True))

df.head()
```

### 2. Lakukan proses encoding pada kolom "Burnout_Risk".


```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

df['Burnout_Risk'] = le.fit_transform(df['Burnout_Risk'])

print('Mapping kelas:')
for kelas, encoded in zip(le.classes_, le.transform(le.classes_)):
    print(f'{kelas} = {encoded}')

# Encoding seluruh variabel kategorikal lainnya
cat_cols = df.select_dtypes(include=['object']).columns
for col in cat_cols:
    df[col] = le.fit_transform(df[col].astype(str))
```

```python
df.head()
```

### 3. Lakukan proses standarisasi pada semua kolom yang memiliki nilai numerik.


```python
# Memisahkan fitur dan target
X = df.drop(columns=['Burnout_Risk'])
y = df['Burnout_Risk']

print('Ukuran X:', X.shape)
print('Ukuran y:', y.shape)
```

```python
# Standardisasi fitur numerik

from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)

X_scaled = pd.DataFrame(
    X_scaled,
    columns=X.columns
)

X_scaled.head()
```

### 4. Lakukan proses stratified split data untuk membuat data latih dan data uji dengan rasio 80:20.


```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_scaled,
    y,
    test_size=0.2,
    random_state=0,
    stratify=y
)
```

```python
print(f'Jumlah data asli: {len(X_scaled)}')
print(f'Jumlah data train: {len(X_train)}')
print(f'Jumlah data test: {len(X_test)}')

print('\nJumlah label data asli:')
print(y.value_counts().sort_index())

print('\nJumlah label data train:')
print(y_train.value_counts().sort_index())

print('\nJumlah label data test:')
print(y_test.value_counts().sort_index())
```
