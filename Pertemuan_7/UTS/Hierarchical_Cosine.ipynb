## 3. Modeling

3.1 Komputasi Matriks Jarak (Distance Matrix)

```python
from scipy.spatial.distance import squareform
from sklearn.metrics import pairwise_distances

X_mat = X_scaled.values

print("Menghitung matriks jarak Cosine...")
t0 = time.perf_counter()

# Matriks jarak berpasangan Cosine N x N (1 - Cosine Similarity)
D_cosine = pairwise_distances(X_mat, metric='cosine')

# Antisipasi rounding error komputasi floating point agar tidak ada nilai < 0
D_cosine = np.clip(D_cosine, 0.0, 2.0)
np.fill_diagonal(D_cosine, 0.0)

durasi_dist = time.perf_counter() - t0
print(f"Selesai dalam {durasi_dist:.2f} detik. Bentuk matriks: {D_cosine.shape}")

# Konversi ke bentuk condensed 1D array untuk SciPy linkage
D_condensed = squareform(D_cosine, checks=False)
```

3.2 Penetapan Jumlah Klaster

```python
from scipy.cluster.hierarchy import linkage, dendrogram

k_target = 3
```

3.3 Pembentukan Hierarki Agglomerative dan Pengukuran Komputasi

```python
tracemalloc.start()
t0 = time.perf_counter()

# Gunakan average linkage pada D_condensed Cosine
Z_agg = linkage(D_condensed, method='average')

durasi_agg = time.perf_counter() - t0
_, peak_mem_agg = tracemalloc.get_traced_memory()
tracemalloc.stop()
peak_mem_agg_mb = peak_mem_agg / (1024 * 1024)
```

3.4 Ekstraksi Label Agglomerative

```python
from scipy.cluster.hierarchy import fcluster

labels_agg = fcluster(Z_agg, t=k_target, criterion='maxclust')
```

3.5 Definisi Algoritma Divisive (DIANA)

```python
def diana_bipartition(indices, dist_matrix):
    indices = np.array(indices, dtype=int)
    n = len(indices)
    if n <= 1:
        return list(indices), []

    sub_d = dist_matrix[np.ix_(indices, indices)]
    sum_cluster = np.sum(sub_d, axis=1)
    avg_dissim = sum_cluster / (n - 1)

    first_local = int(np.argmax(avg_dissim))
    first_splinter = indices[first_local]

    sum_to_splinter = dist_matrix[indices, first_splinter].astype(float)
    sum_to_main = sum_cluster - sum_to_splinter

    in_main_mask = np.ones(n, dtype=bool)
    in_main_mask[first_local] = False

    n_main = n - 1
    n_splinter = 1

    while n_main > 0:
        main_indices_local = np.where(in_main_mask)[0]

        d_main = sum_to_main[main_indices_local] / max(n_main - 1, 1)
        d_splinter = sum_to_splinter[main_indices_local] / n_splinter

        diff = d_main - d_splinter
        best_pos = int(np.argmax(diff))

        if diff[best_pos] <= 0:
            break

        cand_local = main_indices_local[best_pos]
        cand_global = indices[cand_local]

        in_main_mask[cand_local] = False
        n_main -= 1
        n_splinter += 1

        d_cand = dist_matrix[indices, cand_global]
        sum_to_main -= d_cand
        sum_to_splinter += d_cand

    main_group = indices[in_main_mask].tolist()
    splinter_group = indices[~in_main_mask].tolist()

    return main_group, splinter_group

def diana_clustering(dist_matrix, target_k):
    n_samples = dist_matrix.shape[0]
    clusters = [list(range(n_samples))]

    while len(clusters) < target_k:
        diameters = []
        for c in clusters:
            if len(c) > 1:
                sub_d = dist_matrix[np.ix_(c, c)]
                diameters.append(np.max(sub_d))
            else:
                diameters.append(0.0)

        split_idx = int(np.argmax(diameters))
        if diameters[split_idx] == 0:
            break

        cluster_to_split = clusters.pop(split_idx)
        c1, c2 = diana_bipartition(cluster_to_split, dist_matrix)
        clusters.append(c1)
        clusters.append(c2)

    labels = np.zeros(n_samples, dtype=int)
    for c_id, members in enumerate(clusters, start=1):
        labels[members] = c_id
    return labels
```

3.6 Eksekusi DIANA

```python
tracemalloc.start()
t0 = time.perf_counter()

# Eksekusi pemodelan DIANA berbasis matriks jarak Cosine
labels_div = diana_clustering(D_cosine, target_k=k_target)

durasi_div = time.perf_counter() - t0
_, peak_mem_div = tracemalloc.get_traced_memory()
tracemalloc.stop()
peak_mem_div_mb = peak_mem_div / (1024 * 1024)
```

## 4. Evaluasi

4.1 Silhouette Score

```python
sil_agg = silhouette_score(D_cosine, labels_agg, metric='precomputed')
sil_div = silhouette_score(D_cosine, labels_div, metric='precomputed')

print(f"Agglomerative: {sil_agg:.4f}")
print(f"Divisive DIANA: {sil_div:.4f}")
```

4.2 Davies-Bouldin Index

```python
dbi_agg = davies_bouldin_score(X_mat, labels_agg)
dbi_div = davies_bouldin_score(X_mat, labels_div)

print(f"Agglomerative: {dbi_agg:.4f}")
print(f"Divisive DIANA: {dbi_div:.4f}")
```

4.3 Calinski-Harabasz Index

```python
ch_agg = calinski_harabasz_score(X_mat, labels_agg)

ch_div = calinski_harabasz_score(X_mat, labels_div)


print(f"Agglomerative: {ch_agg:.4f}")
print(f"Divisive DIANA: {ch_div:.4f}")
```

4.4 Distribusi Anggota Klaster

```python
distribution = pd.DataFrame({
    "Agglomerative": pd.Series(labels_agg).value_counts().sort_index(),
    "Divisive DIANA": pd.Series(labels_div).value_counts().sort_index(),
}).fillna(0).astype(int)
display(distribution)
```

4.5 Ringkasan Waktu Eksekusi dan Memori

```python
df_runtime = pd.DataFrame({
    "Metode": ["Agglomerative", "Divisive DIANA"],
    "Waktu_Detik": [durasi_agg, durasi_div],
    "Peak_Traced_Memory_MB": [peak_mem_agg_mb, peak_mem_div_mb],
})
display(df_runtime)
print(f"Waktu perhitungan matriks jarak: {durasi_dist:.4f} detik")
```

4.6 Ringkasan Evaluasi

```python
df_eval = pd.DataFrame([
    {
        'Metode': 'Agglomerative',
        'Metrik_Jarak': 'Cosine',
        'Linkage': 'Average',
        'Jumlah_Cluster': k_target,
        'Waktu_Detik': round(durasi_agg, 4),
        'Peak_Memory_MB': round(peak_mem_agg_mb, 2),
        'Silhouette': round(sil_agg, 4),
        'DBI': round(dbi_agg, 4),
        'Calinski_Harabasz': round(ch_agg, 4)
    },
    {
        'Metode': 'Divisive (DIANA)',
        'Metrik_Jarak': 'Cosine',
        'Linkage': '-',
        'Jumlah_Cluster': k_target,
        'Waktu_Detik': round(durasi_div, 4),
        'Peak_Memory_MB': round(peak_mem_div_mb, 2),
        'Silhouette': round(sil_div, 4),
        'DBI': round(dbi_div, 4),
        'Calinski_Harabasz': round(ch_div, 4)
    }
])

display(df_eval)
```

## 5. Visualisasi Hasil

5.1 Visualisasi Dendogram Agglomerative

```python
# Hitung garis potong dinamis untuk k klaster
threshold = (Z_agg[-(k_target - 1), 2] + Z_agg[-k_target, 2]) / 2

plt.figure(figsize=(12, 6))
dendrogram(
    Z_agg,
    truncate_mode='lastp',
    p=30,
    leaf_rotation=90.,
    leaf_font_size=10.,
    show_contracted=True
)
plt.axhline(y=threshold, color='crimson', linestyle='--', label=f'Garis Potong k={k_target} (t={threshold:.2f})')
plt.title('Dendrogram Agglomerative Hierarchical (Cosine - Average Linkage)')
plt.xlabel('Klaster / Titik Sampel (Truncated p=30)')
plt.ylabel('Jarak Cosine')
plt.legend()
plt.tight_layout()
plt.savefig(f'{OUTPUT_DIR}/dendrogram_agg_cosine.png')
plt.show()
```

5.2 Visualisasi Perbandingan 2D (PCA)

```python
pca = PCA(n_components=2, random_state=42)
X_pca = pca.fit_transform(X_mat)

fig, axes = plt.subplots(1, 2, figsize=(14, 6))

scatter1 = axes[0].scatter(X_pca[:, 0], X_pca[:, 1], c=labels_agg, cmap='viridis', alpha=0.6, s=15)
axes[0].set_title(f'Agglomerative (Cosine - k={k_target})')
axes[0].set_xlabel(f'PC1 ({pca.explained_variance_ratio_[0]*100:.1f}%)')
axes[0].set_ylabel(f'PC2 ({pca.explained_variance_ratio_[1]*100:.1f}%)')
plt.colorbar(scatter1, ax=axes[0], label='Label Klaster')

scatter2 = axes[1].scatter(X_pca[:, 0], X_pca[:, 1], c=labels_div, cmap='viridis', alpha=0.6, s=15)
axes[1].set_title(f'Divisive DIANA (Cosine - k={k_target})')
axes[1].set_xlabel(f'PC1 ({pca.explained_variance_ratio_[0]*100:.1f}%)')
axes[1].set_ylabel(f'PC2 ({pca.explained_variance_ratio_[1]*100:.1f}%)')
plt.colorbar(scatter2, ax=axes[1], label='Label Klaster')

plt.suptitle('Proyeksi PCA 2D: Agglomerative vs Divisive (Metrik Cosine)', fontsize=14)
plt.tight_layout()
plt.savefig(f'{OUTPUT_DIR}/pca_comparison_cosine.png')
plt.show()
```
