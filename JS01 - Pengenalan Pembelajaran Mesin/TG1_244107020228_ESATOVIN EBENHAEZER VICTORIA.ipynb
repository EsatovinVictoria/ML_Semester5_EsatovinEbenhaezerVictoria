Berikut merupakan deskripsi dan kegunaan untuk setiap library yang telah diinstall di atas:

PyPREP:
PyPREP adalah library Python yang digunakan untuk melakukan preprocessing pada data EEG (Electroencephalography). Library ini membantu membersihkan data EEG dari berbagai gangguan atau noise, seperti interferensi listrik dan sinyal yang tidak diinginkan, sehingga data menjadi lebih baik untuk digunakan pada tahap analisis maupun pemodelan machine learning. PyPREP juga dapat membantu proses seperti pemeriksaan kualitas data dan penanganan channel EEG yang bermasalah.

SciPy:
SciPy adalah library Python yang digunakan untuk komputasi ilmiah dan teknik. Library ini menyediakan berbagai fungsi yang dapat digunakan untuk matematika, statistik, optimasi, aljabar linear, serta pemrosesan sinyal dan gambar. Dalam pengolahan data, SciPy sering digunakan sebagai pendukung untuk melakukan perhitungan dan analisis yang membutuhkan fungsi-fungsi ilmiah yang lebih kompleks dibandingkan fungsi dasar Python atau NumPy.

wandb:
wandb atau Weights & Biases adalah library yang digunakan untuk melacak, mencatat, dan memvisualisasikan eksperimen machine learning. Dengan wandb, pengguna dapat menyimpan informasi seperti parameter model, nilai loss, akurasi, hasil evaluasi, serta perkembangan proses training. Library ini berguna ketika melakukan beberapa eksperimen karena hasil dari setiap percobaan dapat dibandingkan dan dipantau dengan lebih mudah.

pyECG:
pyECG adalah library Python yang digunakan untuk pengolahan dan analisis sinyal ECG (Electrocardiography). Sinyal ECG merupakan rekaman aktivitas listrik jantung yang dapat digunakan untuk memperoleh informasi mengenai kondisi dan aktivitas jantung. Library ini dapat membantu proses pengolahan sinyal ECG sehingga karakteristik atau informasi penting dari sinyal tersebut dapat dianalisis lebih lanjut, misalnya sebagai bagian dari penelitian atau pengembangan sistem berbasis machine learning.
