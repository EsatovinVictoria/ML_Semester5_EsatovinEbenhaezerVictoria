## Langkah 3 - Inspeksi Visual


```python
# Inspeksi Distribusi Data

num_cols = ['Age', 'Work_Hours_Per_Week']
for col in num_cols:
    plt.figure()
    plt.hist(df[col].dropna(), bins=30)
    plt.title(f'Distribusi {col}')
    plt.xlabel(col); plt.ylabel('Frekuensi')
    plt.show()
```

```python
# Inspeksi Outlier

plt.figure()
plt.boxplot(df['Work_Hours_Per_Week'].dropna(), vert=True)
plt.title('Boxplot Work_Hours_Per_Week')
plt.ylabel('Work_Hours_Per_Week')
plt.show()
```

```python
# Inspeksi Komposisi Variabel Kuantitatif

cat_cols = ['Burnout_Risk', 'Gender', 'Remote_Work', 'Stress_Level']

fig, axes = plt.subplots(2, 2, figsize=(10, 8))
axes = axes.flatten()

for i, col in enumerate(cat_cols):
    df[col].value_counts(dropna=False).plot(kind='bar', ax=axes[i])
    axes[i].set_title(f'Distribusi {col}')
    axes[i].set_xlabel(col)
    axes[i].set_ylabel('Jumlah')

plt.tight_layout()
plt.show()
```

```python
# Inspeksi korelasi Heatmap

import seaborn as sns # Import seaborn for enhanced visualization

num_only = df.select_dtypes(include=[np.number])
corr = num_only.corr(numeric_only=True)

plt.figure(figsize=(14, 12)) # Increased figure size for better readability
sns.heatmap(corr, cmap='coolwarm', vmin=-1, vmax=1, annot=True, fmt=".2f") # Use seaborn.heatmap with annotation
plt.title('Heatmap Korelasi (numerik)')
plt.tight_layout()
plt.show()
```

```python
# Inspeksi Korelasi Scatter Plot

plt.figure()
plt.scatter(df['Work_Hours_Per_Week'], df['Burnout_Score'], alpha=0.7)
plt.xlabel('Work_Hours_Per_Week'); plt.ylabel('Burnout_Score')
plt.title('Scatter Work_Hours_Per_Week vs Burnout_Score')
plt.show()
```
