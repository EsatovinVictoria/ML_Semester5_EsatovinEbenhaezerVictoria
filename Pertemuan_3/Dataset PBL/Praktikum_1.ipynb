## Pra Pengolahan Data - Ekstraksi Fitur


## Langkah 1 - Load Data


```python
import pandas as pd

data = 'burnout_dataset1.csv' # path dataset
df = pd.read_csv(data) # load dataset

df.head()
```

## Langkah 2 - Pengecekan Data


```python
df.info()
```

## Langkah 3 - Imputasi


```python
# Mengisi semua nilai missing sekaligus pada dataframe utama
df.fillna(df.mean(numeric_only=True), inplace=True)
```

```python
df.isnull().sum()
```

## Langkah 4 -  Validasi Hasil


```python
df.info()
```

```python
df.head(10)
```

## Ekstraksi Fitur


```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df["Total_Digital_Hours"] = df["Screen_Time_Hours"] + df["Social_Media_Hours"] + df["Gaming_Hours"]

df["Mental_Strain_Score"] = df["Anxiety_Score"] + df["Depression_Score"]

df["AgeBin"] = pd.cut(df["Age"], bins=[0, 25, 35, 50, 65, 100],
                      labels=["GenZ", "YoungAdult", "Adult", "MiddleAged", "Senior"])

df["Work_To_Sleep_Ratio"] = df["Work_Hours_Per_Week"] / (df["Sleep_Hours"] * 7)

df["Income_Per_Work_Hour"] = df["Monthly_Income_USD"] / (df["Work_Hours_Per_Week"] * 4)

print("Hasil Ekstraksi fitur ")
df[["Age", "AgeBin", "Work_Hours_Per_Week", "Total_Digital_Hours", "Mental_Strain_Score", "Work_To_Sleep_Ratio", "Burnout_Score"]].head()
```

## Langkah 6 - Visualisasi Hasil Ekstraksi Fitur


```python
plt.figure(figsize=(12, 6))
sns.barplot(x="AgeBin", y="Burnout_Score", data=df)
plt.title("Rata-rata Burnout Score berdasarkan Kelompok Usia (AgeBin)")
plt.show()

plt.figure(figsize=(12, 6))
sns.barplot(x="Stress_Level", y="Burnout_Score", data=df,
            order=["Low", "Moderate", "High"])
plt.title("Rata-rata Burnout Score berdasarkan Stress Level")
plt.show()

plt.figure(figsize=(12, 6))
sns.barplot(x="Remote_Work", y="Burnout_Score", data=df,
            order=["No", "Hybrid", "Yes"])
plt.title("Rata-rata Burnout Score berdasarkan Remote Work")
plt.show()

plt.figure(figsize=(12, 6))
sns.barplot(x="Sleep_Quality", y="Burnout_Score", data=df,
            order=["Poor", "Average", "Good", "Excellent"])
plt.title("Rata-rata Burnout Score berdasarkan Sleep Quality")
plt.show()
```
