## 3. Modeling

### 3.1 Komputasi Matriks Jarak (Distance Matrix)

```python
from scipy.spatial.distance import squareform
from sklearn.metrics import pairwise_distances

# Konversi dataframe ke numpy array
X_mat = X_scaled.values

print("Menghitung matriks jarak Manhattan...")
t0 = time.perf_counter()

# Matriks jarak N x N untuk DIANA dan evaluasi
D_manhattan = pairwise_distances(X_mat, metric='manhattan')

durasi_dist = time.perf_counter() - t0
print(f"Selesai dalam {durasi_dist:.2f} detik. Bentuk matriks: {D_manhattan.shape}")

# Konversi ke bentuk condensed 1D array untuk SciPy linkage
D_condensed = squareform(D_manhattan, checks=False)
```

### 3.2 Penetapan Jumlah Klaster

```python
from scipy.cluster.hierarchy import linkage, dendrogram

# Tentukan target k klaster
k_target = 3
```

### 3.3 Pembentukan Hierarki Agglomerative dan Pengukuran Komputasi

```python
tracemalloc.start()
t0 = time.perf_counter()

# Average linkage cocok untuk metrik non-Euclidean
Z_agg = linkage(D_condensed, method='average')

durasi_agg = time.perf_counter() - t0
_, peak_mem_agg = tracemalloc.get_traced_memory()
tracemalloc.stop()
peak_mem_agg_mb = peak_mem_agg / (1024 * 1024)
```

### 3.4 Ekstraksi Label Agglomerative

```python
from scipy.cluster.hierarchy import fcluster

# Potong hierarki ke target k
labels_agg = fcluster(Z_agg, t=k_target, criterion='maxclust')
```

### 3.5 Definisi Algoritma Divisive (DIANA)

```python
def diana_bipartition(indices, dist_matrix):
    """Membagi klaster menggunakan algoritma DIANA yang dioptimasi secara vektor (C-Speed)."""
    indices = np.array(indices, dtype=int)
    n = len(indices)
    if n <= 1:
        return list(indices), []

    # 1. Hitung total jarak internal tiap titik di dalam klaster
    sub_d = dist_matrix[np.ix_(indices, indices)]
    sum_cluster = np.sum(sub_d, axis=1)
    avg_dissim = sum_cluster / (n - 1)

    first_local = int(np.argmax(avg_dissim))
    first_splinter = indices[first_local]

    # Inisialisasi running sum jarak (tanpa loop cand satu per satu)
    sum_to_splinter = dist_matrix[indices, first_splinter].astype(float)
    sum_to_main = sum_cluster - sum_to_splinter

    in_main_mask = np.ones(n, dtype=bool)
    in_main_mask[first_local] = False

    n_main = n - 1
    n_splinter = 1

    # 2. Pindahkan titik ke splinter group dengan operasi vektor
    while n_main > 0:
        main_indices_local = np.where(in_main_mask)[0]

        # Rata-rata jarak titik main ke sesama main dan ke splinter
        d_main = sum_to_main[main_indices_local] / max(n_main - 1, 1)
        d_splinter = sum_to_splinter[main_indices_local] / n_splinter

        diff = d_main - d_splinter
        best_pos = int(np.argmax(diff))

        # Berhenti jika tidak ada lagi titik yang lebih dekat ke splinter group
        if diff[best_pos] <= 0:
            break

        cand_local = main_indices_local[best_pos]
        cand_global = indices[cand_local]

        # Pindahkan titik
        in_main_mask[cand_local] = False
        n_main -= 1
        n_splinter += 1

        # Update running sum secara langsung dalam 1 operasi vektor (O(1))
        d_cand = dist_matrix[indices, cand_global]
        sum_to_main -= d_cand
        sum_to_splinter += d_cand

    main_group = indices[in_main_mask].tolist()
    splinter_group = indices[~in_main_mask].tolist()

    return main_group, splinter_group

def diana_clustering(dist_matrix, target_k):
    """Membagi klaster secara hierarkis sampai mencapai target k."""
    n_samples = dist_matrix.shape[0]
    clusters = [list(range(n_samples))]

    while len(clusters) < target_k:
        diameters = []
        for c in clusters:
            if len(c) > 1:
                sub_d = dist_matrix[np.ix_(c, c)]
                diameters.append(np.max(sub_d))
            else:
                diameters.append(0.0)

        split_idx = int(np.argmax(diameters))
        if diameters[split_idx] == 0:
            break

        cluster_to_split = clusters.pop(split_idx)
        c1, c2 = diana_bipartition(cluster_to_split, dist_matrix)
        clusters.append(c1)
        clusters.append(c2)

    labels = np.zeros(n_samples, dtype=int)
    for c_id, members in enumerate(clusters, start=1):
        labels[members] = c_id
    return labels
```

### 3.6 Eksekusi DIANA

```python
tracemalloc.start()
t0 = time.perf_counter()

# Eksekusi pemodelan DIANA
labels_div = diana_clustering(D_manhattan, target_k=k_target)

durasi_div = time.perf_counter() - t0
_, peak_mem_div = tracemalloc.get_traced_memory()
tracemalloc.stop()
peak_mem_div_mb = peak_mem_div / (1024 * 1024)

# Evaluasi metrik
```

## 4. Evaluasi

### 4.1 Silhouette Score

```python
sil_agg = silhouette_score(D_manhattan, labels_agg, metric='precomputed')
sil_div = silhouette_score(D_manhattan, labels_div, metric='precomputed')

print("=== HASIL EVALUASI AGGLOMERATIVE (MANHATTAN) ===")
print(f"Silhouette Score (Manhattan): {sil_agg:.4f}")
print("=== HASIL EVALUASI DIVISIVE (MANHATTAN) ===")
print(f"Silhouette Score (Manhattan): {sil_div:.4f}")
```

### 4.2 Davies-Bouldin Index

```python
dbi_agg = davies_bouldin_score(X_mat, labels_agg)
dbi_div = davies_bouldin_score(X_mat, labels_div)

print(f"Davies-Bouldin Index       : {dbi_agg:.4f}")
print(f"Davies-Bouldin Index       : {dbi_div:.4f}")
```

### 4.3 Calinski-Harabasz Index

```python
ch_agg = calinski_harabasz_score(X_mat, labels_agg)

ch_div = calinski_harabasz_score(X_mat, labels_div)


print(f"Calinski-Harabasz Score    : {ch_agg:.4f}")
print(f"Calinski-Harabasz Score    : {ch_div:.4f}")
```

### 4.4 Distribusi Anggota Klaster

```python
print(f"Distribusi Anggota Klaster : {pd.Series(labels_agg).value_counts().to_dict()}")
print(f"Distribusi Anggota Klaster : {pd.Series(labels_div).value_counts().to_dict()}")
```

### 4.5 Ringkasan Waktu Eksekusi dan Memori

```python
print(f"Durasi Linkage : {durasi_agg:.4f} detik")
print(f"Peak Memory    : {peak_mem_agg_mb:.2f} MB")
print(f"Waktu Eksekusi             : {durasi_div:.4f} detik")
print(f"Peak Traced Memory         : {peak_mem_div_mb:.2f} MB")
```

### 4.6 Ringkasan Evaluasi

```python
df_eval = pd.DataFrame([
    {
        'Metode': 'Agglomerative',
        'Metrik_Jarak': 'Manhattan',
        'Linkage': 'Average',
        'Jumlah_Cluster': k_target,
        'Waktu_Detik': round(durasi_agg, 4),
        'Peak_Memory_MB': round(peak_mem_agg_mb, 2),
        'Silhouette': round(sil_agg, 4),
        'DBI': round(dbi_agg, 4),
        'Calinski_Harabasz': round(ch_agg, 4)
    },
    {
        'Metode': 'Divisive (DIANA)',
        'Metrik_Jarak': 'Manhattan',
        'Linkage': '-',
        'Jumlah_Cluster': k_target,
        'Waktu_Detik': round(durasi_div, 4),
        'Peak_Memory_MB': round(peak_mem_div_mb, 2),
        'Silhouette': round(sil_div, 4),
        'DBI': round(dbi_div, 4),
        'Calinski_Harabasz': round(ch_div, 4)
    }
])

display(df_eval)

# Simpan tabel metrik evaluasi
```
